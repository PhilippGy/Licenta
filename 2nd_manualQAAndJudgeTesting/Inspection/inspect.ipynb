```python
import faiss
import numpy as np


chunks = np.load("../Data/chunks.npy", allow_pickle=True)
index = faiss.read_index("../Data/index.faiss")
```

```python
# 1. Vezi câte fragmente ai în total
print(f"Total fragmente în baza de date: {len(chunks)}")

# 2. Caută dacă există cuvântul "Informatică" în fragmentele tale
keyword = "Română"
matching_chunks = [c for c in chunks if keyword.lower() in c.lower()]

print(f"Am găsit {len(matching_chunks)} fragmente care conțin '{keyword}':")
for i, m in enumerate(matching_chunks[:3]): # Afișăm doar primele 3
    print(f"\n--- Fragment {i} ---\n{m[:200]}...") # Primele 200 caractere
```

```python
with open("dataset_inspectat.txt", "w", encoding="utf-8") as f:
    for i, chunk in enumerate(chunks):
        f.write(f"CHUNK {i}:\n{chunk}\n{'-'*50}\n")
```
