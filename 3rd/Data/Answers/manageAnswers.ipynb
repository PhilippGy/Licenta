### ANSWER ALL QUESTIONS (replaces answers.json file)

```python
import json
import os
from openai import OpenAI

# Configurare Client
with open("API KEY.txt", "r") as f:
    api_key_str = f.read().strip()
client = OpenAI(api_key=api_key_str)

def generate_gold_answers(input_json, kb_file, output_json):
    # 1. Încărcăm Knowledge Base-ul (Sursa de adevăr)
    try:
        with open(kb_file, 'r', encoding='utf-8') as f:
            kb_content = f.read()
    except FileNotFoundError:
        print(f"Eroare: Nu am găsit fișierul {kb_file}")
        return

    # 2. Încărcăm întrebările
    with open(input_json, 'r', encoding='utf-8') as f:
        questions_data = json.load(f)

    # Definim instrucțiunile (System Prompt)
    system_instructions = """
You are a high-quality ground-truth answer generation model for FMI UBB Cluj admission.
Your task is to generate the official reference answer ("gold answer") in ROMANIAN.

RULES:
- Use ONLY the provided knowledge base.
- If information is missing, state: "Informația nu este specificată în ghidul de admitere."
- If out-of-scope, state: "Această întrebare este în afara ariei de organizare a admiterii FMI."
- Prefer factual correctness over verbosity.
- Language: ROMANIAN ONLY.
"""

    answers_data = []

    for i, item in enumerate(questions_data):
        question = item.get("question")
        
        # Construim mesajul user care conține KB-ul și Întrebarea
        user_message = f"""
KNOWLEDGE BASE (admission_kb.md):
<kb_content>
{kb_content}
</kb_content>

QUESTION:
{question}

Generate the gold answer based strictly on the text inside the <kb_content> tags:
"""

        try:
            response = client.chat.completions.create(
                model="gpt-4.1-mini",
                messages=[
                    {"role": "system", "content": system_instructions},
                    {"role": "user", "content": user_message}
                ],
                temperature=0.0 # 0.0 pentru determinism maxim (răspunsuri stabile)
            )

            answer = response.choices[0].message.content
            
            # Salvăm datele
            new_entry = item.copy()
            new_entry["answer"] = answer
            answers_data.append(new_entry)
            
            print(f"[{i+1}/{len(questions_data)}] Procesat.")

        except Exception as e:
            print(f"Eroare la {i}: {e}")

    # Salvare finală
    with open(output_json, 'w', encoding='utf-8') as f:
        json.dump(answers_data, f, indent=4, ensure_ascii=False)

# if __name__ == "__main__":
#     generate_gold_answers("../Questions/questions.json", "../MD file/admission_kb.md", "answers.json")
```
