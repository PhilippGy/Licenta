## 1. Define the experiment name

```python
EXPERIMENT_NAME = "LLM Wiki Experiments"

# Retrieval
TOP_K = 20
RERANKED_K = 8  # set to 0 to disable reranking

# To change the dataset, refer to administrative.ipynb.
```

## 2. Generate the questions

```python
import json

# Define the file paths
input_txt_file = "generateQuestions.txt"
output_json_file = f"Experiments/{EXPERIMENT_NAME}/questions.json"

# Initialize a list to hold the question dictionaries
questions_list = []

# Open and read the text file line by line
with open(input_txt_file, 'r', encoding='utf-8') as file:
    for line in file:
        # Strip leading/trailing whitespace (like \n newline characters)
        cleaned_line = line.strip()
        
        # Check to ensure we don't add empty lines to the JSON
        if cleaned_line:
            # Create a dictionary for the current line and add it to the list
            questions_list.append({"question": cleaned_line})

# Save the list of dictionaries into a JSON file
with open(output_json_file, 'w', encoding='utf-8') as json_file:
    # indent=4 makes the JSON file human-readable (pretty-printed)
    # ensure_ascii=False allows non-English characters to save properly
    json.dump(questions_list, json_file, indent=4, ensure_ascii=False)

print(f"Successfully converted '{input_txt_file}' to '{output_json_file}'!")
```

## 3. Run experiments

```python
import sys
import os
from sentence_transformers import SentenceTransformer
from sentence_transformers import CrossEncoder
from functools import partial
import faiss
import numpy as np
from pathlib import Path

parent_dir = os.path.abspath(os.path.join(os.getcwd(), '..'))
if parent_dir not in sys.path:
    sys.path.append(parent_dir)

from Functionality.answerExperimentalQuestions import answerAndScore
from Functionality.retrieval import retrieve



SCRIPT_DIR = Path.cwd()

CHUNKS_PATH = SCRIPT_DIR.parent / "Data/" / "chunks.npy"
INDEX_PATH = SCRIPT_DIR.parent / "Data/" / "index.faiss"

chunks = np.load(CHUNKS_PATH, allow_pickle=True)
index = faiss.read_index(str(INDEX_PATH))


embedding_model = SentenceTransformer("intfloat/multilingual-e5-small")
reranker_model = CrossEncoder("cross-encoder/mmarco-mMiniLMv2-L12-H384-v1")
retrieval = partial(retrieve, embedding_model, index, chunks, reranker_model)

answerAndScore(retrieve=retrieval, path=EXPERIMENT_NAME, initialK=TOP_K, finalK=RERANKED_K)
```
