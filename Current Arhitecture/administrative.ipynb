## BUILD A NEW KNOWLEDGE BASE
- Run the code block below to generate a new markdown file containing the admission website data as of today.
- To check whether the currently active MD file is outdated, check the log.txt located in Data/Rebuilding_the_data to see when the code has last been run.
- To change the current knowledge base to the newly generated MD file, go to the next section further below.

```python
from datetime import datetime
from Data.Rebuilding_the_data.crawl import fetch_all_docs
from Data.Rebuilding_the_data.Markdown_conversion.convertDocsToMarkdown import build_knowledge_base



# current API calling method requires a model from OpenAI
MODEL = "gpt-4.1-mini"   

# Log the time of the iteration
current_time = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
with open("Data/Rebuilding_the_data/log.txt", "a") as file:
    file.write(f"{current_time} - {MODEL}\n")

fetch_all_docs()   # crawls the university admissions website
build_knowledge_base()   # frontier model generates a markdown file admission_kb based on crawled text
```
