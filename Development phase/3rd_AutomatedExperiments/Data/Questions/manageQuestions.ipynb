### REMOVE QUESTION SETS BY QUESTION/CATEGORY/TYPE

```python
import json
import os

file_name = "questions.json"

# Define the type you want to remove
field_to_filter_by = "type"
remove_by_field_name = "hallucination trap" 

if os.path.exists(file_name):
    # 1. Read the existing data
    with open(file_name, 'r', encoding='utf-8') as f:
        try:
            data_list = json.load(f)
        except json.JSONDecodeError:
            print("Error: Could not read JSON file.")
            data_list = []

    # 2. Filter the list
    # This keeps every item UNLESS its type matches type_to_remove
    original_count = len(data_list)
    filtered_list = [item for item in data_list if item.get(field_to_filter_by) != remove_by_field_name]
    
    removed_count = original_count - len(filtered_list)

    # 3. Write the filtered list back to the file
    with open(file_name, 'w', encoding='utf-8') as f:
        json.dump(filtered_list, f, indent=4)

    print(f"Done! Removed {removed_count} items of {field_to_filter_by} '{remove_by_field_name}'.")
else:
    print(f"File {file_name} not found.")
```

### CHECK QUESTION SET BY NUMBER 

```python
import json
import os

file_name = "questions.json"


if os.path.exists(file_name):
    with open(file_name, 'r', encoding='utf-8') as f:
        try:
            data_list = json.load(f)
            
            while True:
                user_input = input("Enter question number to check (or 'q' to quit): ")
                if user_input.lower() == 'q':
                    break
                
                try:
                    # The question number you want to check (starting from 1)
                    target_number = int(user_input)
                    
                    # Check if the number is within the range of the list
                    if 1 <= target_number <= len(data_list):
                        # Access the item using index (number - 1)
                        item = data_list[target_number - 1]
                        
                        print(f"--- Details for Question #{target_number} ---")
                        print(f"Question: {item.get('question')}")
                        print(f"Type:     {item.get('type')}")
                        print(f"Category: {item.get('category')}")
                        print("-" * 35)
                    else:
                        print(f"Error: Question #{target_number} not found. (Total questions: {len(data_list)})")
                        
                except ValueError:
                    print("Please enter a valid number.")
            
        except json.JSONDecodeError:
            print("Error: Could not parse JSON. The file might be corrupted or empty.")
else:
    print(f"File '{file_name}' not found.")
```

### REMOVE QUESTION SET BY NUMBER

```python
import json
import os

file_name = "questions.json"
# The question number you want to remove (starting from 1)
target_number = -1

if os.path.exists(file_name):
    with open(file_name, 'r', encoding='utf-8') as f:
        try:
            data_list = json.load(f)
        except json.JSONDecodeError:
            print("Error: JSON file is empty or invalid.")
            data_list = []

    # Check if the number is valid for the current list
    if 1 <= target_number <= len(data_list):
        # target_number - 1 converts human counting to 0-based index
        removed_item = data_list.pop(target_number - 1)
        
        # Save the updated list
        with open(file_name, 'w', encoding='utf-8') as f:
            json.dump(data_list, f, indent=4)
        
        print(f"Successfully deleted question #{target_number}: '{removed_item.get('question')[:30]}...'")
    else:
        print(f"Error: Question #{target_number} does not exist. Total count is {len(data_list)}.")
else:
    print(f"File {file_name} not found.")
```

### COUNT TOTAL AMOUNT OF QUESTION SETS

```python
import json
import os

file_name = "questions.json"

if os.path.exists(file_name):
    with open(file_name, 'r', encoding='utf-8') as f:
        try:
            data = json.load(f)
            # Check if the data is a list before counting
            if isinstance(data, list):
                print(f"Total questions in file: {len(data)}")
            else:
                print("The JSON file does not contain a list.")
        except json.JSONDecodeError:
            print("The file is empty or contains invalid JSON.")
else:
    print(f"File '{file_name}' not found.")
```
